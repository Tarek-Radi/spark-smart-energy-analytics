```python
df = spark.read.csv("data/NullData.csv", header=True, inferSchema=True)
```

```python
df.show(truncate=False)
```

```python
df.na.drop().show(truncate=False)
```

```python
df.na.drop(thresh=2).show(truncate=False)  
#thresh=2 means that if a row has at least 2 non-null values, it will be kept; otherwise, it will be dropped.
```

```python
df.show(truncate=False)
```

```python
df.na.drop(subset=['Name','Sales']).show(truncate=False)  
#subset=['Name','Sales'] means that only the specified columns will be considered when determining whether to drop a row. If a row has null values in either the 'Name' or 'Sales' column, it will be dropped; otherwise, it will be kept.
```

```python
df.na.fill({"Name": "Unknown", "Sales": 0}).show(truncate=False)
# fill() method is used to replace null values in the DataFrame with specified values. In this case, it replaces null values in the 'Name' column with "Unknown" and null values in the 'Sales' column with 0.
```

```python
m= df.select(mean(col("Sales")).alias("Average_Sales")).show(truncate=False)
```

```python
m= df.select(mean(col("Sales")).alias("Average_Sales")).collect()
print(m[0]["Average_Sales"])
```
